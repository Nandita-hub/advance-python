```python
from collections import OrderedDict

class LRUCache:
    def __init__(self, capacity):
        self.cache = OrderedDict()
        self.cap = capacity

    def get(self, key):
        if key not in self.cache:
            return -1
        self.cache.move_to_end(key)
        return self.cache[key]

    def put(self, key, value):
        if key in self.cache:
            self.cache.move_to_end(key)
        self.cache[key] = value
        if len(self.cache) > self.cap:
            removed = self.cache.popitem(last=False)
            print(" Removed least recently used:", removed)

    def display(self):
        print("Current Cache:", dict(self.cache))


capacity = int(input("Enter cache capacity: "))
lru = LRUCache(capacity)

while True:
    print("\n===== LRU CACHE MENU =====")
    print("1. Put (Add Key-Value)")
    print("2. Get (Access Key)")
    print("3. Display Cache")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        key = input("Enter key: ")
        value = input("Enter value: ")
        lru.put(key, value)
        print("Added/Updated")

    elif choice == "2":
        key = input("Enter key: ")
        result = lru.get(key)
        if result == -1:
            print(" Key not found")
        else:
            print(" Value:", result)

    elif choice == "3":
        lru.display()

    elif choice == "4":
        print(" Exiting...")
        break

    else:
        print(" Invalid choice!")
```

```python
def power(base, exp):
    result = 1
    for _ in range(exp):
        result *= base
    return result

print(power(2, 3))
```

```python
s = input("Enter string: ")

vowels = "aeiouAEIOU"
v = c = d = sp = 0

for ch in s:
    if ch in vowels:
        v += 1
    elif ch.isalpha():
        c += 1
    elif ch.isdigit():
        d += 1
    else:
        sp += 1

print("Palindrome:", s == s[::-1])
print(v, c, d, sp)
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"Student Name: {self.name}, Marks: {self.marks}")


class Teacher:
    def __init__(self, name, subject):
        self.name = name
        self.subject = subject

    def display(self):
        print(f"Teacher Name: {self.name}, Subject: {self.subject}")


class Admin:
    def __init__(self):
        self.students = []
        self.teachers = []

    def add_student(self, name, marks):
        s = Student(name, marks)
        self.students.append(s)
        print(" Student added")

    def add_teacher(self, name, subject):
        t = Teacher(name, subject)
        self.teachers.append(t)
        print(" Teacher added")

    def show_students(self):
        if not self.students:
            print("No students found")
        else:
            for s in self.students:
                s.display()

    def show_teachers(self):
        if not self.teachers:
            print("No teachers found")
        else:
            for t in self.teachers:
                t.display()

    def manage(self):
        print(" Admin managing system...")


admin = Admin()

while True:
    print("\n===== SCHOOL MANAGEMENT =====")
    print("1. Add Student")
    print("2. Add Teacher")
    print("3. Show Students")
    print("4. Show Teachers")
    print("5. Admin Manage")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        name = input("Enter student name: ")
        marks = int(input("Enter marks: "))
        admin.add_student(name, marks)

    elif choice == "2":
        name = input("Enter teacher name: ")
        subject = input("Enter subject: ")
        admin.add_teacher(name, subject)

    elif choice == "3":
        admin.show_students()

    elif choice == "4":
        admin.show_teachers()

    elif choice == "5":
        admin.manage()

    elif choice == "6":
        print(" Exiting...")
        break

    else:
        print(" Invalid choice!")
```

```python
class Logger:
    def __init__(self, filepath):
        self.file = open(filepath, "a")
        print(" Log file opened")

    def log(self, msg, level="INFO"):
        self.file.write(f"{level}: {msg}\n")

    def log_info(self, msg):
        self.log(msg, "INFO")

    def log_warning(self, msg):
        self.log(msg, "WARNING")

    def log_error(self, msg):
        self.log(msg, "ERROR")

    def __del__(self):
        print("Closing log file...")
        self.file.close()

file_name = input("Enter log file name: ")
logger = Logger(file_name)

while True:
    print("\n===== LOGGER MENU =====")
    print("1. Log Info")
    print("2. Log Warning")
    print("3. Log Error")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        msg = input("Enter message: ")
        logger.log_info(msg)
        print("Info logged")

    elif choice == "2":
        msg = input("Enter message: ")
        logger.log_warning(msg)
        print("Warning logged")

    elif choice == "3":
        msg = input("Enter message: ")
        logger.log_error(msg)
        print("Error logged")

    elif choice == "4":
        print("Exiting...")
        del logger   # triggers destructor
        break

    else:
        print(" Invalid choice!")
```

```python
class Vehicle:
    count = 0

    def __init__(self, name):
        self.name = name
        Vehicle.count += 1

    def rent(self, days):
        return 0


class Car(Vehicle):
    def rent(self, days):
        return 1000 * days


class Bike(Vehicle):
    def rent(self, days):
        return 500 * days

vehicles = []

while True:
    print("\n===== VEHICLE RENTAL SYSTEM =====")
    print("1. Rent Car")
    print("2. Rent Bike")
    print("3. Show Total Vehicles Rented")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter customer name: ")
        days = int(input("Enter number of days: "))

        car = Car(name)
        vehicles.append(car)

        cost = car.rent(days)
        print(f" Car rented by {name}")
        print("Total cost:", cost)

    elif choice == "2":
        name = input("Enter customer name: ")
        days = int(input("Enter number of days: "))

        bike = Bike(name)
        vehicles.append(bike)

        cost = bike.rent(days)
        print(f"Bike rented by {name}")
        print("Total cost:", cost)

    elif choice == "3":
        print(" Total Vehicles Rented:", Vehicle.count)

    elif choice == "4":
        print(" Exiting system...")
        break

    else:
        print(" Invalid choice!")
```

```python
class User:
    def __init__(self, name):
        self.name = name


class Comment:
    def __init__(self, text):
        self.text = text


class Post:
    count = 0

    def __init__(self, content):
        self.content = content
        self.likes = 0
        self.comments = []
        Post.count += 1

    def like(self):
        self.likes += 1

    def add_comment(self, comment):
        self.comments.append(comment)

user_name = input("Enter your name: ")
user = User(user_name)

posts = []

while True:
    print("\n===== MENU =====")
    print("1. Create Post")
    print("2. Like Post")
    print("3. Comment on Post")
    print("4. View Posts")
    print("5. Exit")

    choice = input("Enter your choice: ")


    if choice == "1":
        content = input("Enter post content: ")
        post = Post(content)
        posts.append(post)
        print(" Post created!")

    elif choice == "2":
        if not posts:
            print(" No posts available!")
            continue

        for i, p in enumerate(posts):
            print(i, ":", p.content)

        try:
            idx = int(input("Select post number: "))
            if 0 <= idx < len(posts):
                posts[idx].like()
                print(" Post liked!")
            else:
                print(" Invalid post number!")
        except:
            print(" Enter a valid number!")

    elif choice == "3":
        if not posts:
            print(" No posts available!")
            continue

        for i, p in enumerate(posts):
            print(i, ":", p.content)

        try:
            idx = int(input("Select post number: "))
            if 0 <= idx < len(posts):
                text = input("Enter comment: ")
                comment = Comment(text)
                posts[idx].add_comment(comment)
                print("Comment added!")
            else:
                print(" Invalid post number!")
        except:
            print(" Enter a valid number!")

    elif choice == "4":
        if not posts:
            print("No posts yet!")
        else:
            for i, p in enumerate(posts):
                print(f"\n--- Post {i} ---")
                print("Content:", p.content)
                print("Likes:", p.likes)
                print("Comments:")
                if p.comments:
                    for c in p.comments:
                        print("-", c.text)
                else:
                    print("No comments")

    # Exit
    elif choice == "5":
        print("\nTotal Posts Created:", Post.count)
        print(" Exiting program...")
        break

    else:
        print(" Invalid choice! Please try again.")
```
