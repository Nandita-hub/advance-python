```python
import re

text = "Hello Python"
pattern = "Python"

if re.search(pattern, text):
    print("Pattern found")
else:
    print("Not found")
```

```python
#1. Largest element in list
import re
data = input("Enter numbers: ")
nums = list(map(int, re.findall(r'\d+', data)))
print("Largest:", max(nums))
```

```python
#2. Even or Odd
import re
num = input("Enter number: ")
if re.fullmatch(r'\d+', num):
    print("Even" if int(num)%2==0 else "Odd")
```

```python
#3. Reverse string
import re
s = input("Enter string: ")
rev = ""
for ch in s:
    rev = ch + rev
print("Reverse:", rev)
```

```python
#4. Count vowels
import re
s = input("Enter string: ")
print("Vowels:", len(re.findall(r'[aeiouAEIOU]', s)))
```

```python
#5. Sum of digits
import re
num = input("Enter number: ")
digits = re.findall(r'\d', num)
print("Sum:", sum(map(int, digits)))
```

```python
#6. Palindrome
import re
s = input("Enter string: ")
clean = ''.join(re.findall(r'\w', s)).lower()
print("Palindrome" if clean==clean[::-1] else "Not")
```

```python
#7. Fibonacci
import re
n = int(input("Enter n: "))
a,b=0,1
for _ in range(n):
    print(a,end=" ")
    a,b=b,a+b
```

```python
#10. Min & Max
import re
nums = list(map(int, re.findall(r'\d+', input("Enter: "))))
print("Min:", min(nums), "Max:", max(nums))
```

```python
#11. Second largest
import re
nums = list(set(map(int, re.findall(r'\d+', input("Enter: ")))))
nums.sort()
print("Second largest:", nums[-2])
```

```python
#12. Remove duplicates
import re
nums = list(map(int, re.findall(r'\d+', input("Enter: "))))
res=[]
for i in nums:
    if i not in res:
        res.append(i)
print(res)
```

```python
#13. Frequency
import re
nums = re.findall(r'\d+', input("Enter: "))
freq={}
for i in nums:
    freq[i]=freq.get(i,0)+1
print(freq)
```

```python
#14. Merge sorted lists
import re
a = list(map(int, re.findall(r'\d+', input("List1: "))))
b = list(map(int, re.findall(r'\d+', input("List2: "))))
print(sorted(a+b))
```

```python
#15. Intersection
import re
a = list(map(int, re.findall(r'\d+', input("List1: "))))
b = list(map(int, re.findall(r'\d+', input("List2: "))))
print([i for i in a if i in b])
```

```python
#16. Rotate list
import re
nums = list(map(int, re.findall(r'\d+', input("Enter: "))))
k = int(input("k: "))
print(nums[k:]+nums[:k])
```

```python
#17. Anagram
import re
s1 = ''.join(re.findall(r'[a-z]', input().lower()))
s2 = ''.join(re.findall(r'[a-z]', input().lower()))
print("Anagram" if sorted(s1)==sorted(s2) else "Not")
```

```python
#18. First non-repeating
import re
s = re.findall(r'[a-z]', input().lower())
for ch in s:
    if s.count(ch)==1:
        print(ch)
        break
```

```python
#19. Flatten list
lst = [[1,2],[3,4]]
flat=[]
for sub in lst:
    for i in sub:
        flat.append(i)
print(flat)
```

```python
#20. Pairs with sum
import re
nums = list(map(int, re.findall(r'\d+', input("Enter: "))))
target = int(input("Target: "))
for i in range(len(nums)):
    for j in range(i+1,len(nums)):
        if nums[i]+nums[j]==target:
            print(nums[i], nums[j])
```

```python
#21. Right triangle
n=int(input())
for i in range(1,n+1):
    print("*"*i)
```

```python
#22. Pyramid
n=int(input())
for i in range(n):
    print(" "*(n-i-1)+"*"*(2*i+1))
```

```python
#23. Prime numbers
start=int(input())
end=int(input())
for num in range(start,end+1):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```

```python
#24. Armstrong
import re
num = input("Enter: ")
digits = re.findall(r'\d', num)
p = len(digits)
total = sum(int(d)**p for d in digits)
print("Armstrong" if total==int(num) else "Not")
```
